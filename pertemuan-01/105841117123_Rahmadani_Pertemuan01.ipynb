```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.datasets import load_wine

%matplotlib inline
```

```python
# ============================================================
# 1. NumPy — operasi dasar array
# ============================================================

x = np.array([1, 2, 3, 4, 5])

print("=== NumPy ===")
print("Array:", x)
print("Mean :", x.mean())
print("Std  :", x.std())
```

```python
# ============================================================
# 2. Pandas — DataFrame sederhana
# ============================================================

df = pd.DataFrame({
    "nama": ["A", "B", "C"],
    "nilai": [80, 75, 90]
})

print("=== Pandas ===")
display(df)
```

```python
# ============================================================
# 3. Visualisasi — bar chart
# ============================================================

plt.figure()
plt.bar(df["nama"], df["nilai"])
plt.title("Nilai Mahasiswa")
plt.xlabel("Nama")
plt.ylabel("Nilai")
plt.show()
```

```python
# ============================================================
# 4. Dataset ML — load Iris
# ============================================================

iris = load_iris(as_frame=True)

print("=== Dataset Iris (5 baris pertama) ===")
display(iris.frame.head())
```

```python
# ============================================================
# Tugas 1: array random 20 angka, hitung mean, median, std
# ============================================================

np.random.seed(42)
arr = np.random.rand(20) * 100

print("=== Tugas 1: Statistik Array Random ===")
print("Array :", np.round(arr, 1))
print("Mean  :", round(arr.mean(), 2))
print("Median:", round(np.median(arr), 2))
print("Std   :", round(arr.std(), 2))
```

```python
# ============================================================
# Tugas 2: DataFrame 10 mahasiswa, cari 3 nilai tertinggi
# ============================================================

np.random.seed(0)

df_mhs = pd.DataFrame({
    "nama": [f"Mahasiswa_{i+1}" for i in range(10)],
    "nilai": np.random.randint(50, 100, 10)
})

print("Data Nilai Mahasiswa:")
display(df_mhs)

top3 = df_mhs.nlargest(3, "nilai")

print("Top 3 Nilai:")
display(top3)
```

```python
# ============================================================
# Tugas 3: 2 grafik dari data nilai
# ============================================================

# Grafik 1: Bar Chart
plt.figure()
plt.bar(df_mhs["nama"], df_mhs["nilai"])
plt.title("Bar Chart Nilai Mahasiswa")
plt.xticks(rotation=45)
plt.show()

# Grafik 2: Histogram
plt.figure()
plt.hist(df_mhs["nilai"], bins=5)
plt.title("Histogram Nilai Mahasiswa")
plt.show()
```

```python
# ============================================================
# Tugas 4: Load dataset Wine
# ============================================================

wine = load_wine(as_frame=True)

print("Ukuran data:", wine.frame.shape)
print("Kolom:", list(wine.frame.columns))

print("\nInsight 1: Dataset Wine memiliki",
      wine.frame.shape[0], "sampel dan",
      wine.frame.shape[1] - 1, "fitur.")

print("Insight 2: Target memiliki",
      wine.frame["target"].nunique(),
      "kelas:", sorted(wine.frame["target"].unique()))

print("Insight 3: Rata-rata alcohol =",
      round(wine.frame["alcohol"].mean(), 2))

print("\n✅ Pertemuan 01 selesai.")
```
