```python
"""
Pertemuan 5: Support Vector Machine (SVM)
=========================================
"""

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report
```

```python
# ============================================================
# 1. Load & Split Dataset
# ============================================================
data = load_breast_cancer()

X_train, X_test, y_train, y_test = train_test_split(
    data.data,
    data.target,
    test_size=0.2,
    random_state=42
)
```

```python
# ============================================================
# 2. Scaling 
# ============================================================
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# ============================================================
# 3. Tugas 1: Bandingkan Kernel Linear vs RBF
# ============================================================
print("=" * 60)
print("Tugas 1: Perbandingan Kernel Linear vs RBF")
print("=" * 60)

models = {
    "SVM Linear": SVC(kernel="linear", C=1),
    "SVM RBF": SVC(kernel="rbf", C=1, gamma="scale")
}

results = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    acc = round(accuracy_score(y_test, y_pred), 3)
    results[name] = acc

    print(f"\n{name}")
    print("-" * 40)
    print(f"Accuracy: {acc}")
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred, target_names=data.target_names))
```

```python
# ============================================================
# 4. Tugas 2: Tuning C dan gamma (RBF)
# ============================================================
print("\n" + "=" * 60)
print("Tugas 2: Tuning Parameter C dan gamma (Kernel RBF)")
print("=" * 60)

best_acc = 0
best_params = {}

for c_val in [0.1, 1, 10]:
    for gamma_val in [0.01, 0.1, 1]:
        model = SVC(kernel="rbf", C=c_val, gamma=gamma_val)
        model.fit(X_train, y_train)
        acc = accuracy_score(y_test, model.predict(X_test))

        print(f"C={c_val:<4} | gamma={gamma_val:<4} → accuracy: {acc:.3f}")

        if acc > best_acc:
            best_acc = acc
            best_params = {"C": c_val, "gamma": gamma_val}

print("\nParameter Terbaik:")
print(f"C = {best_params['C']}")
print(f"gamma = {best_params['gamma']}")
print(f"Accuracy = {best_acc:.3f}")
```

```python
# ============================================================
# 5. Tugas 3: Kesimpulan
# ============================================================
print("\n" + "=" * 60)
print("KESIMPULAN")
print("=" * 60)

print("1. Kernel RBF umumnya memberikan performa lebih baik dibandingkan")
print("   kernel Linear pada dataset Breast Cancer karena mampu")
print("   menangkap pola non-linear.")

print("2. Parameter C dan gamma sangat mempengaruhi performa model.")
print(f"   Kombinasi terbaik pada percobaan ini adalah C={best_params['C']} "
      f"dan gamma={best_params['gamma']} dengan accuracy {best_acc:.3f}.")

print("3. Scaling menggunakan StandardScaler wajib dilakukan sebelum")
print("   menggunakan SVM karena model ini sensitif terhadap perbedaan skala fitur.")

print("\n✅ Pertemuan 05 selesai.")
```
